## Model architecture

The model is a stacked LSTM (two LSTM layers, strictly left-to-right /
causal) -- the standard choice for next-word / text generation. It gives
enough capacity for longer-range dependencies while remaining causal,
which matches exactly how the model is used at inference time:
predicting one word at a time from everything generated so far, never
from future context. Layer sizes and dropout are configurable via
`ModelParams` in `params.yaml`.
